# Project 3: SQL Data Analysis

The following SQL queries satisfy SELECT, WHERE, ORDER BY, GROUP BY, COUNT, SUM and AVG requirements.

```
-- Display all records
SELECT * FROM Orders;

-- Delivered orders
SELECT * FROM Orders WHERE OrderStatus='Delivered';

-- Sort by total price
SELECT OrderID, Product, TotalPrice
FROM Orders
ORDER BY TotalPrice DESC;

-- Product counts
SELECT Product, COUNT(*) AS TotalOrders
FROM Orders
GROUP BY Product;

-- Average order value
SELECT AVG(TotalPrice) AS AverageOrderValue
FROM Orders;

-- Total revenue
SELECT SUM(TotalPrice) AS TotalRevenue
FROM Orders;
```
